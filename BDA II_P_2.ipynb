## 6.2.2 Which word is occuring the most?

This must be done in three steps. Reason is, that the vectorizer.vocabulary_ is organized as a dictonary with the value indicating the position of the word in the array
1. Find out the highest count of a word
2. Find out the position of this count
3. Find out the word at this position

### Find out the highest count of a word

The variable `numOccurencesPerWord` is an array which contains the number of occurrences of each word. 

The [max()](https://pandas.pydata.org/docs/reference/api/pandas.Series.max.html) function finds the highest value in the array and stores this value in a new variable `highest_count_of_word`.

### Find out the word at this position

By using the [idmax()](https://pandas.pydata.org/docs/reference/api/pandas.Series.idxmax.html) function, I return the word with the highest occurrence. 

In this case the row label is the word I am searching for, and the maximum value is the highest count of a word

```python
word_with_highest_occurence = numOccurencesPerWord.idxmax()
print("\nWord with the highest occurence:", word_with_highest_occurence, "\n")
```

The word with the highest number of occurrences is the word `the` with a number of 7975 counts. I even expected this result, because `the` is one of the stop words that occur too often

### Find out the position of this count

After that, I use the variable in which the word with the highest number is stored as the key for the [vocabulary_](https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html) function and display the position of the word. 

```python
position_of_word = vocabulary_[word_with_highest_occurence]
print("\nPosition of the word:", position_of_word, "\n")
```

# 6.4 Rescaling the data using term frequency inverse document frequency
Here, term frequency is the number of occurences of a term (word) $t$ in a document $d$. 

$\operatorname{tf}(t, d) = f_{t, d}$ 

Sometimes tf gets normalized to the length of $d$
The inverse document frequency idf is a measure on the amount of information a term t carries. Rare occurences of t leads to a high amount of information common occurence to a low amount of information. The idf is computed as 

$\text{idf}(t) = \log{\frac{1 + n}{1+\text{df}(t)}} + 1$

where $n$ is the total number of documents and $\text{df}(t)$ is the number of documents that contain the term $t$. Hence, the tf-idf is the product of the two terms:

$\text{tf-idf(t,d)}=\text{tf(t,d)} \cdot \text{idf(t)}$

scikit-learn supports this in the `TfidfTransformer`, when using the following parameters: `TfidfTransformer(norm='l2', use_idf=True, smooth_idf=True, sublinear_tf=False)`. Refer to the scikit documentation for the parameter sets and how this changes the formula.

Combining Bag of Words and tf-idf can be done using the `TfidfVectorizer`

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vect = TfidfVectorizer(input="filename", ngram_range=(2,2),stop_words="english", max_df=0.75, min_df=3)
X_tfidf = pd.DataFrame(tfidf_vect.fit_transform(filenames).toarray(), columns=tfidf_vect.get_feature_names_out())
X_tfidf
```

Each word now possesses a TF-IDF value when it appears in a document, or zero if it is absent in a document (row of the matrix).

# 6.4.1 Find maximum value for each of the features over dataset

In this section, we will find the maximum value for each feature in the dataset.

We generate a single value for each term, indicating its significance in the classification of a document

To make it a little clearer and more obvious, I have displayed the first 20 and the last 20 features with their maximum values to make the difference in value clearer.

```python
# your code here
max_values = X_tfidf.max(axis=0)
sorted_by_tfidf = max_values.argsort()
feature_names = X_tfidf.columns
print("Features with lowest tfidf:\n{}".format(
    feature_names[sorted_by_tfidf[:20]]))
print("\ntf-idf values:\n{}".format(max_values.iloc[sorted_by_tfidf[:20]]))
print("\nFeatures with highest tfidf: \n{}".format(
    feature_names[sorted_by_tfidf[-20:]]))
print("\ntf-idf values:\n{}".format(max_values.iloc[sorted_by_tfidf[-20:]]))
```

##### Features with lowest tf-idf:

The words listed have low TF-IDF values, which indicates that they occur frequently in the documents analysed and have a low information across all documents. They could be everyday words or stop words that are less important for understanding the text. In our case, stop words were removed so that represent typical phrases in the entire text corpus or are not very relevant with regard to classify a document.

##### Features mit höchstem tf-idf:

Words with high TF-IDF values are less common throughout the corpus, and are words that indicate high information.

##### General conclusions:

Words like "holmes examined" and "mean say" exhibit low TF-IDF values, suggesting that they are common phrases across the entire text corpus. The phrase "night went" stands out with a slightly higher TF-IDF value among the lower values, implying that this word pairing occurs less frequently and may offer more nuanced context.
