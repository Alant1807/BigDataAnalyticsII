## Detect Missing Values

To find out which missing data I had to deal with, I tried to load the dataset into a dataframe in advance and convert all columns except the last column to floats. Then I tried to output the dataframe and found that in some lines the string '?' could not be converted. This gave me the hint that these must be missing values.

To find out which rows and columns are affected by missing values, I have defined a method `rows_with_missing_Values` that outputs all rows with missing values marked with '?'.

I applied the method to the dataframe with [dataframe.apply](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html) so I knew which columns were affected by missing values. The rows which have missing data are rows 6,9,13,14,18 and all columns have at least one missing value.

Since pandas seem to be unable to determine the correct data type when there is missing data, each column has the data type `object`, which will be fixed in the next step.

```python
filename = r"D:\datasets\iris\data_someMissing.all"
```

```python
df = pd.read_csv(filename, sep=' ', header=None)

df.columns = ['sl', 'sw', 'pl', 'pw', 'class']

def rows_with_missing_Values(row):
    for value in row:
        if '?' in value:
            return True
    return False

missingData_rows = df[df.apply(rows_with_missing_Values, axis=1)]

print("\nRows with missing values:\n")
print(missingData_rows, "\n")
print(df.dtypes, "\n")
print(df.describe(), "\n")
print(df.info(),"\n")
print(df)
```

## Determine the proper type of columns

First of all, a CSV file named `filename` is loaded into a Pandas DataFrame. The file uses spaces as separators between columns. We disable the use of the first line as column name (header) with `header=None`, and we mark "?" as missing values to detect them in the running process. By using the parameter `na_values` I was able to mark all `?` as NaN, so that pandas was able to determine the proper type of columns by itself. The proper type of the columns sl, sw, pl and pw are float64 and for the class column the proper type is object.

```python
df = pd.read_csv(filename, sep=' ', header=None, na_values=['?'], na_filter=True)

df.columns = ['sl', 'sw', 'pl', 'pw', 'class']

rows_with_NaN = df.iloc[[6, 9, 13, 14, 18]]


print(df.isnull().sum(), "\n")
print(rows_with_NaN,"\n")
print(df,"\n")
print(df.dtypes, "\n")
print(df.describe(), "\n")
print(df.info())
```

## What are your options to work with the missing values? 
* When I have a dataset, I would always want to know at the beginning if my dataset has missing data. In pandas, `NaN` is the default missing value marker. In my use case, I had to mark the missing data as `NaN` in order for pandas to recognise the missing data. After that, there are several options on how to work with missing data.
* First option would be to remove the lines where missing values occurs with [dropna()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.dropna.html?highlight=dropna).
* Second option would be filling missing values using [fillna()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.fillna.html), [replace()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.replace.html) or [interpolate()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.interpolate.html). I can replace the missing value with a scalar value or using a PandasObject like the mean of a specific column.

## Checking for missing values using isna()

I start by checking for missing values in the DataFrame and displaying the first 20 rows to get an initial overview of which columns have missing data. The `null_values` DataFrame will contain True for missing values and False for non-missing values in the first 20 rows.

Next, I calculate and display statistics for all features in the DataFrame. This provides me with information such as the count, mean, standard deviation, minimum, and maximum values for each numeric feature. The `feature_stats` DataFrame will include statistics for all the numeric columns in the original DataFrame.

Finally, I group the data by the 'class' column and calculate statistics for each class separately. This allows me to analyze the distribution of features within each class. The `class_stats` DataFrame provides statistics for each class, helping me understand how the data varies between different classes.

Because we only checked for `NaN` values, the statistical values of each feature and class did not change.

```python
null_values = df.isna().head(20)

feature_stats = df.describe()

class_stats = df.groupby('class').describe()
```

```python
# print statistics for each of the options you realize
print("DataFrame with NaN Values (Boolean):\n")
print(df)

print("\nStatistical Values for Each Feature:\n")
print(feature_stats)
print("\nStatistical Values for Each Class:\n")
print(class_stats)
```

## Dropping missing values using dropna()

I remove rows with missing values from the DataFrame, resulting in a new DataFrame `df_cleaned`. The `df_cleaned` DataFrame now contains the original data with all rows containing missing values removed.

just like before i print the statistics each feature and each class. In this case the class `Iris-setosa` has lost values, because 5 rows got deleted with values which were connected with the class `Iris-setosa`. 

```python
df_cleaned = df.dropna()

feature_stats = df_cleaned.describe()

class_stats = df_cleaned.groupby('class').describe()
```

```python
print("\nDataFrame after Removing Rows with NaN Values:\n")
print(df_cleaned.head(20),"\n")
print(df_cleaned)

print("\nStatistical Values for Each Feature:\n")
print(feature_stats)
print("\nStatistical Values for Each Class:\n")
print(class_stats)
```

## Filling missing values using fillna()

I address missing values in the DataFrame by specify a forward-fill (ffill) to propagate the previous value in a forward direction, resulting in a new DataFrame called `df_filled`. The `df_filled` DataFrame now contains the original data with missing values replaced by the previous value. I chose this method because it fills the missing values in this case with semantically meaningful values.

Like before only the class `Iris-setosa` is affected with the statistical change because the missing values appears in class `Iris-setosa`. 

```python
df_filled = df.fillna(method='ffill')

feature_stats = df_filled.describe()

class_stats = df_filled.groupby('class').describe()
```

```python
print("\nDataFrame after Filling NaN Values:\n")
print(df_filled.head(20))

print("\nStatistical Values for Each Feature:\n")
print(feature_stats)
print("\nStatistical Values for Each Class:\n")
print(class_stats)
```

## Filling missing values using replace()

I address missing values in the DataFrame by replacing them with the mean value of their respective columns. This results in a new DataFrame called `df_replaced`. The `df_replaced` DataFrame now contains the original data with missing values replaced by the mean value of their respective columns.

```python
df_replaced = df.replace(np.nan, df.mean(numeric_only=True))

feature_stats = df_replaced.describe()

class_stats = df_replaced.groupby('class').describe()
```

```python
print("\nDataFrame after Replacing NaN Values with Mean:\n")
print(df_replaced.head(20))

print("\nStatistical Values for Each Feature:\n")
print(feature_stats)
print("\nStatistical Values for Each Class:\n")
print(class_stats)
```
